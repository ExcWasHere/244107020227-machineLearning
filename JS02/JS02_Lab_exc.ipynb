```python
df.head()
```

```python
not_usable= ['id', 'Unnamed: 32']
df = df.drop(columns=not_usable)
```

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
numeric_features = df.select_dtypes(include=np.number).columns
a = StandardScaler()
df[numeric_features] = a.fit_transform(df[numeric_features])

df[numeric_features].agg(['mean', 'std'])
```
